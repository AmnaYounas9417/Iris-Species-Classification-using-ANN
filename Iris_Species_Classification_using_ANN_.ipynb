```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
```

```python
# Extract features (x) and target labels (y)
x = df.drop(columns=['species'])

# Encode the target 'species' strings into integer labels (y_int)
le = LabelEncoder()
y_int = le.fit_transform(df['species'])

# Split the dataset into training and testing sets
xtrain, xtest, ytrain, ytest = train_test_split(x, y_int, test_size=0.2, random_state=42, stratify=y_int)
```

```python
scaler = StandardScaler()
xtrain_scaled = scaler.fit_transform(xtrain)
xtest_scaled = scaler.transform(xtest)
```

```python
per = Perceptron(max_iter=1000, random_state=42)
per.fit(xtrain_scaled, ytrain)
```

```python
y_pred_perceptron = per.predict(xtest_scaled)
```

```python
accuracy =accuracy_score(ytest, y_pred_perceptron)
print(f"Accuracy: {accuracy}")
```

```python
print(classification_report(ytest, y_pred_perceptron))
```

```python
x_train_cat=to_categorical(ytrain,num_classes=3)
print(x_train_cat)
```

```python
y_test_cat = to_categorical(ytest, num_classes=3)
print(y_test_cat)
```

```python
model = Sequential(
    [
        Dense(16, activation='relu', input_shape=(4,)),
        Dense(8, activation='relu'),
        Dense(3, activation='softmax')
    ]
)
```

```python
model.compile (optimizer='adam',loss='categorical_crossentropy',metrics = ['accuracy'])
```

```python
history = model.fit(xtrain_scaled, x_train_cat, epochs=100, batch_size=8, validation_split=0.2, verbose=1)
```

```python
loss,acc=model.evaluate(xtest_scaled,y_test_cat,verbose=1)
print("Loss:",loss)
print("Accuracy:",acc)
```

```python
plt.figure(figsize=(10, 4))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(history.history['accuracy'], label='Training accuracy')
plt.plot(history.history['val_accuracy'], label='val_Training accuracy')
```
